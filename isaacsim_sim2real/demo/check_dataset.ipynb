#### Merge dataset

```python
repo_id_1 = f"{USER}/test"
dataset_1 = LeRobotDataset(
    repo_id=repo_id_1,
    root=f"{DATASET_ROOT_PATH}/{repo_id_1}",
)

repo_id_2 = f"{USER}/test_real_simple"
dataset_2 = LeRobotDataset(
    repo_id=repo_id_2,
    root=f"{DATASET_ROOT_PATH}/{repo_id_2}",
)
```

```python
merged_repo_id = f"{USER}/merged_test"
merge_datasets(
    datasets=[dataset_1, dataset_2],
    output_repo_id=merged_repo_id,
    output_dir=f"{DATASET_ROOT_PATH}/{merged_repo_id}",
)
```

```python
load_merged_dataset = LeRobotDataset(
    repo_id=merged_repo_id,
    root=f"{DATASET_ROOT_PATH}/{merged_repo_id}",
)
```

```python
print(f"dataset 1 total frames: {dataset_1.meta.total_frames}")
print(f"dataset 2 total frames: {dataset_2.meta.total_frames}")
print(f"merged dataset total frames: {load_merged_dataset.meta.total_frames}")
print(dataset_1.meta.total_frames + dataset_2.meta.total_frames == load_merged_dataset.meta.total_frames)
```
